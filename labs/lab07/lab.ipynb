Funkcje w `exercises.py` zwracają **tekst zapytania SQL**; test uruchamia go na tabelach `orders` i `customers`. SQL jest przenośny: ten sam kod działa w SQLite i DuckDB.

```python
import sqlite3
sales = pd.read_csv(DATA / "sales.csv"); cust = pd.read_csv(DATA / "customers.csv")
conn = sqlite3.connect(":memory:")
sales.to_sql("orders", conn, index=False); cust.to_sql("customers", conn, index=False)
def q(sql): return pd.read_sql_query(sql, conn)
q("SELECT country, COUNT(*) n, ROUND(SUM(amount),2) revenue FROM orders GROUP BY country ORDER BY revenue DESC")
```
