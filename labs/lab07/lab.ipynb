## Cele
Matplotlib (obiektowe API `fig, ax`), dobór wykresu do pytania, czytelność (tytuły, osie, jednostki).

```python
import matplotlib.pyplot as plt
sales = pd.read_csv(DATA / "sales.csv", parse_dates=["date"])
monthly = sales.groupby(sales["date"].dt.strftime("%Y-%m"))["amount"].sum()
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(monthly.index, monthly.values); ax.tick_params(axis="x", rotation=45)
ax.set_title("Przykład"); plt.show()
```

**Dobór wykresu:** trend w czasie → linia; porównanie kategorii → słupki; rozkład → histogram/box; zależność dwóch zmiennych → scatter.
