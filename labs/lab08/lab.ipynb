## Część A – wizualizacja i EDA

```python
import matplotlib.pyplot as plt
sales = pd.read_csv(DATA / "sales.csv", parse_dates=["date"])
monthly = sales.groupby(sales["date"].dt.strftime("%Y-%m"))["amount"].sum()
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(monthly.index, monthly.values); ax.tick_params(axis="x", rotation=45); ax.set_title("Przykład"); plt.show()
```

**Dobór wykresu:** trend w czasie → linia · porównanie kategorii → słupki · rozkład → histogram/box · zależność → scatter. Uczciwa wizualizacja: oś Y słupków od 0, opisane osie, tytuł-wniosek.

## Część B – niepewność
Czy klienci z PL i DE różnią się średnią kwotą zamówienia?

```python
pl = sales.loc[sales.country == "PL", "amount"].to_numpy(); de = sales.loc[sales.country == "DE", "amount"].to_numpy()
print(len(pl), len(de), pl.mean().round(1), de.mean().round(1))
```
