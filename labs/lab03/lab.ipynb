## Rozgrzewka: pętla vs wektor

```python
import time
x = np.random.default_rng(0).normal(size=2_000_000)

t = time.perf_counter(); s1 = sum(v * v for v in x); t1 = time.perf_counter() - t
t = time.perf_counter(); s2 = float(x @ x);           t2 = time.perf_counter() - t
print(f"pętla: {t1:.3f}s  wektor: {t2:.4f}s  przyspieszenie ≈ {t1 / t2:.0f}x")
```

## Kształty i osie
`axis=0` agreguje **wzdłuż wierszy** (wynik per kolumna), `axis=1` – per wiersz.

```python
A = np.arange(12).reshape(3, 4)
print(A.shape, A.sum(axis=0), A.sum(axis=1))
print("broadcasting:", (A - A.mean(axis=0)).shape)
```
