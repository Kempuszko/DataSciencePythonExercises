## Cele
Data leakage – najczęstszy błąd w ML. Feature engineering i pipeline'y, które chronią przed przeciekiem.

```python
churn = pd.read_csv(DATA / "churn.csv")
churn.groupby("cancellation_reason")["churned"].mean()
```

Kolumna `cancellation_reason` jest znana **dopiero po** odejściu klienta – to ewidentny przeciek. Model, który jej użyje,
osiągnie niemal 100% i będzie bezużyteczny w produkcji.

**Typowe źródła przecieku:** (1) cechy z przyszłości, (2) skalowanie/imputacja na całych danych przed splitem,
(3) target encoding bez separacji train/test, (4) duplikaty między train a test.

## Eksperyment: ile "daje" przeciek?

```python
try:
    from sklearn.linear_model import LogisticRegression
    from sklearn.model_selection import cross_val_score
    base = pd.get_dummies(churn[["tenure_months", "support_calls", "contract"]], drop_first=True).astype(float)
    leak = base.join(pd.get_dummies(churn["cancellation_reason"]).astype(float))
    for name, X in [("uczciwe cechy", base), ("z przeciekiem", leak)]:
        auc = cross_val_score(LogisticRegression(max_iter=2000), X, churn["churned"], cv=5, scoring="roc_auc").mean()
        print(f"{name:15s} AUC = {auc:.3f}")
except ImportError:
    print("Zainstaluj scikit-learn")
```
