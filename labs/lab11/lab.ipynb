## Cele
Metryki klasyfikacji (nie tylko accuracy), ROC/AUC, walidacja krzyżowa — implementacja od zera.

```python
churn = pd.read_csv(DATA / "churn.csv")
print("odsetek odejść:", churn["churned"].mean().round(3))
print("accuracy 'zawsze 0':", (1 - churn["churned"].mean()).round(3))
```

Model "zawsze przewiduj 0" ma wysoką accuracy, a jest bezużyteczny. Dlatego potrzebujemy precision/recall/F1/AUC.

## Zadania (autograding)
`confusion_counts`, `precision_recall_f1`, `roc_auc_manual`, `kfold_indices`.

## Użycie z modelem (po zrobieniu zadań)

```python
try:
    from sklearn.linear_model import LogisticRegression
    X = pd.get_dummies(churn[["tenure_months", "support_calls", "contract"]], drop_first=True).astype(float)
    y = churn["churned"].to_numpy()
    cut = 1500
    clf = LogisticRegression(max_iter=1000).fit(X.iloc[:cut], y[:cut])
    proba = clf.predict_proba(X.iloc[cut:])[:, 1]
    print("AUC (sklearn):", __import__("sklearn.metrics", fromlist=["x"]).roc_auc_score(y[cut:], proba))
except ImportError:
    print("Zainstaluj scikit-learn")
```
