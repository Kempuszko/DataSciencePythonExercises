## Część A – regresja liniowa

```python
house = pd.read_csv(DATA / "housing.csv")
house.describe().T
```

### Zadania A (autograding): `train_test_split_manual`, `fit_linear`, `predict_linear`, `r2_score_manual`
Porównanie ze scikit-learn (po zadaniach):

```python
try:
    from sklearn.linear_model import LinearRegression
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import r2_score
    X = house.drop(columns="price_pln"); y = house["price_pln"]
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=0)
    lr = LinearRegression().fit(Xtr, ytr)
    print("sklearn R2 (test):", round(r2_score(yte, lr.predict(Xte)), 3))
    print(dict(zip(X.columns, lr.coef_.round(0))), lr.intercept_.round(0))
except ImportError:
    print("Zainstaluj scikit-learn (conda env update -f environment.yml)")
```

## Część B – klasyfikacja i metryki

```python
churn = pd.read_csv(DATA / "churn.csv")
print("odsetek odejść:", churn["churned"].mean().round(3), "| accuracy 'zawsze 0':", (1 - churn["churned"].mean()).round(3))
```

Model „zawsze przewiduj 0” ma wysoką accuracy, a jest bezużyteczny – stąd precision/recall/F1.
