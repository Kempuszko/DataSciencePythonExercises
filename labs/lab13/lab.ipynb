## Eksperyment: drzewo vs las (po zrobieniu zadań)

```python
try:
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.tree import DecisionTreeClassifier
    from sklearn.model_selection import cross_val_score
    churn = pd.read_csv(DATA / "churn.csv")
    X = pd.get_dummies(churn[["tenure_months", "support_calls", "monthly_charges", "contract"]].fillna(70),
                       drop_first=True).astype(float)
    y = churn["churned"]
    for m in [DecisionTreeClassifier(random_state=0), DecisionTreeClassifier(max_depth=4, random_state=0),
              RandomForestClassifier(200, random_state=0)]:
        print(type(m).__name__, getattr(m, "max_depth", None),
              cross_val_score(m, X, y, cv=5, scoring="roc_auc").mean().round(3))
except ImportError:
    print("Zainstaluj scikit-learn")
```
