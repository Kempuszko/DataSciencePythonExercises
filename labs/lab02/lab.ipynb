## Część A – kod z AI: czy jest dobrze?
Typowa odpowiedź asystenta AI na prompt „średnia krocząca”. Znajdź błędy **zanim** uruchomisz.

```python
def moving_average_ai(xs, k):
    out = []
    for i in range(len(xs) - k):
        out.append(sum(xs[i:i + k]) / k)
    return out

print(moving_average_ai([1, 2, 3, 4], 2))   # oczekujemy [1.5, 2.5, 3.5]
```

**Checklista przeglądu kodu z AI:** off-by-one · przypadki brzegowe (pusta lista, k=0, k>len) · typy zwracane · mutacje wejścia · wydajność · halucynacje (czy ta funkcja/parametr naprawdę istnieje?).

**Testy jako specyfikacja** – zanim poprosisz AI o kod, zapisz oczekiwania:

```python
def test_moving_average_spec(f):
    assert f([1, 2, 3, 4], 2) == [1.5, 2.5, 3.5]
    assert f([5], 1) == [5.0]
    try:
        f([1, 2], 3); raise AssertionError("brak ValueError")
    except ValueError:
        pass

try:
    test_moving_average_spec(moving_average_ai)
except AssertionError as e:
    print("Kod z AI nie spełnia specyfikacji:", repr(e))
```
