## Część A – AI napisało to za Ciebie. Czy jest dobrze? (15 min)
Poniższa funkcja to typowa odpowiedź asystenta AI na prompt "średnia krocząca". Znajdź błędy **zanim** uruchomisz.

```python
def moving_average_ai(xs, k):
    out = []
    for i in range(len(xs) - k):
        out.append(sum(xs[i:i + k]) / k)
    return out

print(moving_average_ai([1, 2, 3, 4], 2))   # oczekujemy [1.5, 2.5, 3.5]
```

**Checklista przeglądu kodu z AI** (używaj jej w całym kursie):
1. *Off-by-one* – granice pętli/zakresów, indeksy.
2. *Przypadki brzegowe* – pusta lista, k=0, k>len, None, NaN.
3. *Typy zwracane* – lista vs ndarray vs Series, int vs float.
4. *Mutacje wejścia* – czy funkcja psuje argument?
5. *Wydajność* – pętle tam, gdzie wystarczy wektoryzacja.
6. *Halucynacje* – czy funkcje/parametry biblioteki naprawdę istnieją? (sprawdź w dokumentacji)

## Część B – Testy jako specyfikacja (15 min)
Zanim poprosisz AI o kod, zapisz oczekiwania jako `assert`:

```python
def test_moving_average_spec(f):
    assert f([1, 2, 3, 4], 2) == [1.5, 2.5, 3.5]
    assert f([5], 1) == [5.0]
    try:
        f([1, 2], 3); raise AssertionError("brak ValueError")
    except ValueError:
        pass

try:
    test_moving_average_spec(moving_average_ai)
except AssertionError as e:
    print("Kod z AI nie spełnia specyfikacji:", repr(e))
```
